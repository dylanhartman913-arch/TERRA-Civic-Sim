# 08a — Ecoregion × BA Crosswalk

Builds the spatial crosswalk between EPA Level III ecoregions and Balancing Authority territories
in the Mountain West study area. Assigns each synthetic bus to an ecoregion.

**Outputs:**
- `data/processed/mw_ecoregions.geojson` — Mountain West Level III polygons
- `data/processed/ecoregion_ba_crosswalk.geojson` — intersection fragments with area
- `data/processed/figures/ecoregion_ba_overlay.html` — Folium overlay map
- Updated `data/processed/network_metadata.json` — ecoregion_layer block

## Cell 1 — Imports and schema print

```python
import json, warnings
from pathlib import Path

import folium
import geopandas as gpd
import numpy as np
import pandas as pd
from shapely.geometry import box

warnings.filterwarnings('ignore', category=FutureWarning)
warnings.filterwarnings('ignore', message='.*CRS.*')

ROOT = Path('/Users/dylanhartman/Library/CloudStorage/OneDrive-UniversityofWyoming/Research/Energy Modeling/energy-map')

# ── Load all three spatial files ──────────────────────────────────────────────
ba_gdf = gpd.read_file(ROOT / 'data/processed/ba_territories.geojson')
buses_gdf = gpd.read_file(ROOT / 'data/processed/synthetic_buses.geojson')
eco_raw = gpd.read_file(ROOT / 'data/raw/ecoregions/us_eco_l3_state_boundaries.shp')
with open(ROOT / 'data/processed/network_metadata.json') as f:
    meta = json.load(f)

print('=== ba_territories.geojson ===')
print(f'  shape : {ba_gdf.shape}')
print(f'  CRS   : {ba_gdf.crs}')
print(f'  cols  : {list(ba_gdf.columns)}')
print()

print('=== synthetic_buses.geojson ===')
print(f'  shape : {buses_gdf.shape}')
print(f'  CRS   : {buses_gdf.crs}')
print(f'  cols  : {list(buses_gdf.columns)}')
print(f'  roles : {buses_gdf["role"].value_counts().to_dict()}')
print()

print('=== EPA Level III ecoregion shapefile ===')
print(f'  shape : {eco_raw.shape}')
print(f'  CRS   : {eco_raw.crs}')
print(f'  cols  : {list(eco_raw.columns)}')
print()
print(f'  US_L3CODE unique values (n={eco_raw["US_L3CODE"].nunique()}):')
for code, name in sorted(eco_raw.drop_duplicates('US_L3CODE').set_index('US_L3CODE')['US_L3NAME'].items()):
    print(f'    {str(code):>4}  {name}')

print()
print('=== network_metadata.json top-level keys ===')
for k, v in meta.items():
    print(f'  {k}: {str(v)[:80]}')
```

## Cell 2 — Clip ecoregions to Mountain West bounding box

```python
# Mountain West study area bounding box
MW_BBOX = dict(lon_min=-115, lat_min=36, lon_max=-100, lat_max=49)
bbox_4326 = box(MW_BBOX['lon_min'], MW_BBOX['lat_min'], MW_BBOX['lon_max'], MW_BBOX['lat_max'])
bbox_gdf_4326 = gpd.GeoDataFrame(geometry=[bbox_4326], crs='EPSG:4326')

# Reproject bbox to match ecoregion CRS for clipping, then reproject result to 4326
bbox_eco_crs = bbox_gdf_4326.to_crs(eco_raw.crs)

# Clip (uses shapely intersection under the hood)
eco_clipped = gpd.clip(eco_raw, bbox_eco_crs)

# Reproject to EPSG:4326
eco_mw = eco_clipped.to_crs('EPSG:4326')[['US_L3CODE', 'US_L3NAME', 'NA_L3CODE', 'NA_L3NAME',
                                           'NA_L2CODE', 'NA_L2NAME', 'STATE_NAME', 'geometry']].copy()

# Summarise what's in the clipped layer
summary = (
    eco_mw.groupby(['US_L3CODE', 'US_L3NAME'])
    .size()
    .reset_index(name='n_polygons')
    .sort_values('US_L3CODE')
)
print(f'Ecoregions in Mountain West clip ({len(summary)} unique Level-III types, {len(eco_mw)} polygon fragments):')
print(summary.to_string(index=False))

# Save
out_mw = ROOT / 'data/processed/mw_ecoregions.geojson'
eco_mw.to_file(out_mw, driver='GeoJSON')
print(f'\nSaved: {out_mw}  ({out_mw.stat().st_size/1e3:.1f} KB)')
```

## Cell 3 — Spatial intersection: ecoregion × BA territory

```python
# Keep only BA territories that overlap the MW bbox
ba_mw = gpd.clip(ba_gdf[['ba_code', 'ba_name', 'geometry']].copy(), bbox_gdf_4326)
print(f'BA territories overlapping MW bbox: {len(ba_mw)}')

# Reproject both to EPSG:5070 (Albers Equal Area) for accurate area calculation
eco_5070 = eco_mw[['US_L3CODE', 'US_L3NAME', 'geometry']].to_crs('EPSG:5070')
ba_5070 = ba_mw.to_crs('EPSG:5070')

# Spatial overlay — intersection
print('Computing spatial intersection (ecoregion × BA)...')
xwalk_5070 = gpd.overlay(eco_5070, ba_5070, how='intersection', keep_geom_type=True)
xwalk_5070 = xwalk_5070[xwalk_5070.geometry.is_valid & ~xwalk_5070.geometry.is_empty].copy()

# Compute area in km²
xwalk_5070['area_km2'] = xwalk_5070.geometry.area / 1e6

# Drop slivers < 1 km²
xwalk_5070 = xwalk_5070[xwalk_5070['area_km2'] >= 1.0].copy()

# Reproject back to EPSG:4326 for storage
xwalk = xwalk_5070.to_crs('EPSG:4326').rename(columns={
    'US_L3CODE': 'ecoregion_code',
    'US_L3NAME': 'ecoregion_name',
})

print(f'\nCrossalk fragments: {len(xwalk)}')
print(f'Unique ecoregion × BA combinations: {xwalk[["ecoregion_code","ba_code"]].drop_duplicates().shape[0]}')

# Top 15 combinations by area
top = (
    xwalk.groupby(['ecoregion_code', 'ecoregion_name', 'ba_code', 'ba_name'])
    ['area_km2'].sum()
    .reset_index()
    .sort_values('area_km2', ascending=False)
    .head(15)
)
print('\nTop 15 ecoregion × BA combinations by area (km²):')
print(top[['ecoregion_code','ecoregion_name','ba_code','area_km2']].to_string(index=False))

# Save
out_xwalk = ROOT / 'data/processed/ecoregion_ba_crosswalk.geojson'
xwalk.to_file(out_xwalk, driver='GeoJSON')
print(f'\nSaved: {out_xwalk}  ({out_xwalk.stat().st_size/1e3:.1f} KB)')
```

## Cell 4 — Point-in-polygon: assign buses to ecoregions

```python
# Spatial join: buses → ecoregion
# Buses outside the MW clip area will get NaN (left join preserves all buses)
buses_joined = gpd.sjoin(
    buses_gdf,
    eco_mw[['US_L3CODE', 'US_L3NAME', 'geometry']].rename(columns={
        'US_L3CODE': 'ecoregion_code',
        'US_L3NAME': 'ecoregion_name',
    }),
    how='left',
    predicate='within'
).drop(columns=['index_right'])

# Handle duplicates from buses landing on polygon borders
buses_joined = buses_joined.drop_duplicates(subset=['bus_id'], keep='first')

# Restore None for unassigned
n_assigned = buses_joined['ecoregion_code'].notna().sum()
n_total = len(buses_joined)
n_unassigned = n_total - n_assigned

print(f'Bus ecoregion assignment:')
print(f'  Total buses      : {n_total}')
print(f'  Assigned         : {n_assigned}  ({100*n_assigned/n_total:.1f}%)')
print(f'  Unassigned       : {n_unassigned}')

if n_unassigned > 0:
    print('\n  Unassigned buses by BA:')
    print(buses_joined[buses_joined['ecoregion_code'].isna()]
          .groupby('ba_code').size().sort_values(ascending=False)
          .to_string())

# Summary table: ecoregion × BA — bus count and generation capacity
assigned = buses_joined[buses_joined['ecoregion_code'].notna()].copy()
summary_tbl = (
    assigned.groupby(['ecoregion_code', 'ecoregion_name', 'ba_code'])
    .agg(
        bus_count=('bus_id', 'count'),
        total_gen_mw=('generation_cap_mw', 'sum')
    )
    .reset_index()
    .sort_values(['ecoregion_code', 'ba_code'])
)
print('\nEcoregion × BA summary (assigned buses only):')
print(summary_tbl.to_string(index=False))
```

## Cell 5 — Folium overlay map

```python
# Color palette for Level III ecoregions (categorical, up to ~20 in MW)
ECO_PALETTE = [
    '#a6cee3','#1f78b4','#b2df8a','#33a02c','#fb9a99','#e31a1c',
    '#fdbf6f','#ff7f00','#cab2d6','#6a3d9a','#ffff99','#b15928',
    '#8dd3c7','#ffffb3','#bebada','#fb8072','#80b1d3','#fdb462',
    '#b3de69','#fccde5',
]

# Assign a color to each ecoregion
unique_eco_codes = sorted(eco_mw['US_L3CODE'].unique())
eco_color = {code: ECO_PALETTE[i % len(ECO_PALETTE)] for i, code in enumerate(unique_eco_codes)}

# Role colors for buses (match notebook 07 convention)
ROLE_COLORS = {
    'load': '#e41a1c',
    'generation': '#4daf4a',
    'substation': '#377eb8',
    'load_generation': '#ff7f00',
    'load_generation_substation': '#984ea3',
    'generation_substation': '#a65628',
}

# Build map
m = folium.Map(
    location=[42.5, -108.0],
    zoom_start=5,
    tiles='CartoDB positron',
    width='100%', height='100%',
)

# ── Layer 1: Ecoregion polygons ───────────────────────────────────────────────
eco_layer = folium.FeatureGroup(name='EPA Level III Ecoregions', show=True)
for _, row in eco_mw.iterrows():
    color = eco_color.get(row['US_L3CODE'], '#cccccc')
    folium.GeoJson(
        row['geometry'].__geo_interface__,
        style_function=lambda x, c=color: {
            'fillColor': c,
            'color': c,
            'weight': 0.5,
            'fillOpacity': 0.20,
        },
        tooltip=folium.Tooltip(f"{row['US_L3CODE']} — {row['US_L3NAME']}"),
    ).add_to(eco_layer)
eco_layer.add_to(m)

# ── Layer 2: BA territory boundaries (dashed) ────────────────────────────────
ba_layer = folium.FeatureGroup(name='BA Territory Boundaries', show=True)
for _, row in ba_mw.iterrows():
    folium.GeoJson(
        row['geometry'].__geo_interface__,
        style_function=lambda x: {
            'fillColor': 'none',
            'color': '#222222',
            'weight': 1.5,
            'dashArray': '6 4',
            'fillOpacity': 0,
        },
        tooltip=folium.Tooltip(row['ba_code']),
    ).add_to(ba_layer)
ba_layer.add_to(m)

# ── Layer 3: Synthetic buses ──────────────────────────────────────────────────
# Only Mountain West buses (those assigned an ecoregion + their WACM/PACE peers)
MW_BAS = set(xwalk['ba_code'].unique())
buses_mw = buses_joined[buses_joined['ba_code'].isin(MW_BAS)].copy()

bus_layer = folium.FeatureGroup(name='Synthetic Buses', show=True)
for _, b in buses_mw.iterrows():
    role = str(b.get('role', 'load'))
    color = ROLE_COLORS.get(role, '#888888')
    cap_mw = float(b.get('generation_cap_mw', 0) or 0)
    # Radius: 3–12 px proportional to log(cap+1)
    radius = 3 + 9 * (np.log1p(cap_mw) / np.log1p(5000))
    radius = float(np.clip(radius, 3, 12))
    eco_label = b.get('ecoregion_name') or 'Outside MW clip'
    folium.CircleMarker(
        location=[b.geometry.y, b.geometry.x],
        radius=radius,
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.65,
        weight=0.8,
        tooltip=folium.Tooltip(
            f"Bus {b['bus_id']} | BA: {b['ba_code']} | Role: {role}<br>"
            f"Gen: {cap_mw:.0f} MW | Ecoregion: {eco_label}"
        ),
    ).add_to(bus_layer)
bus_layer.add_to(m)

# ── Legend ────────────────────────────────────────────────────────────────────
legend_items_eco = ''.join(
    f'<div><span style="display:inline-block;width:14px;height:14px;background:{eco_color[c]};'
    f'opacity:0.5;margin-right:4px;"></span>{c} — {eco_mw[eco_mw["US_L3CODE"]==c]["US_L3NAME"].iloc[0]}</div>'
    for c in unique_eco_codes
)
legend_items_role = ''.join(
    f'<div><span style="display:inline-block;width:10px;height:10px;border-radius:50%;background:{clr};'
    f'margin-right:4px;"></span>{role}</div>'
    for role, clr in ROLE_COLORS.items()
)
legend_html = f'''
<div style="position:fixed;bottom:30px;left:30px;z-index:9999;background:white;
            border:1px solid #ccc;border-radius:4px;padding:10px;font-size:11px;max-height:400px;overflow-y:auto;">
  <b>EPA Level III Ecoregions</b><br>{legend_items_eco}
  <br><b>Bus Role</b><br>{legend_items_role}
  <br><small>Circle size ∝ log(generation capacity)</small>
</div>
'''
m.get_root().html.add_child(folium.Element(legend_html))

folium.LayerControl(collapsed=False).add_to(m)

out_map = ROOT / 'data/processed/figures/ecoregion_ba_overlay.html'
m.save(str(out_map))
print(f'Saved: {out_map}  ({out_map.stat().st_size/1e3:.1f} KB)')

# Show inline (truncated for large notebooks)
m
```

## Cell 6 — Update network_metadata.json

```python
# Compute Mountain West bus assignment coverage
mw_ba_codes = set(xwalk['ba_code'].unique())
buses_in_mw = buses_gdf[buses_gdf['ba_code'].isin(mw_ba_codes)]
buses_in_mw_joined = buses_joined[buses_joined['ba_code'].isin(mw_ba_codes)]
assigned_mw = buses_in_mw_joined['ecoregion_code'].notna().sum()
coverage_pct = 100.0 * assigned_mw / len(buses_in_mw) if len(buses_in_mw) > 0 else 0.0

# Build ecoregion_layer block
eco_list = (
    eco_mw.drop_duplicates('US_L3CODE')[['US_L3CODE', 'US_L3NAME']]
    .sort_values('US_L3CODE')
    .rename(columns={'US_L3CODE': 'code', 'US_L3NAME': 'name'})
    .to_dict(orient='records')
)

meta['ecoregion_layer'] = {
    'n_ecoregions_in_study_area': len(eco_list),
    'ecoregions': eco_list,
    'bus_assignment_coverage_pct': round(coverage_pct, 2),
    'n_mw_buses_assigned': int(assigned_mw),
    'n_mw_buses_total': int(len(buses_in_mw)),
    'study_area_bbox': [MW_BBOX['lon_min'], MW_BBOX['lat_min'], MW_BBOX['lon_max'], MW_BBOX['lat_max']],
    'crosswalk_fragments': int(len(xwalk)),
    'crosswalk_unique_combinations': int(xwalk[['ecoregion_code','ba_code']].drop_duplicates().shape[0]),
}

with open(ROOT / 'data/processed/network_metadata.json', 'w') as f:
    json.dump(meta, f, indent=2)

print('Updated network_metadata.json with ecoregion_layer block:')
print(json.dumps(meta['ecoregion_layer'], indent=2))
```
